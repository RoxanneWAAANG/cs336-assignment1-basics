```python
for lr in [1e1, 1e2, 1e3]:
    weights = torch.nn.Parameter(5 * torch.randn((10, 10)))
    opt = SGD([weights], lr=lr)

    print(f"\nLearning rate = {lr}")

    for t in range(10):
        opt.zero_grad()

        loss = (weights ** 2).mean()

        loss.backward()
        opt.step()

        print(f"step {t}: loss = {loss.item()}")
```

I ran the toy SGD example for 10 iterations with learning rates 1e1, 1e2, and 1e3. With lr = 1e1, the loss decreases faster than with a small learning rate, although it may be less smooth. With lr = 1e2, the loss diverges: instead of decreasing, it increases over training. With lr = 1e3, the divergence is even more severe, and the loss quickly becomes extremely large or numerically unstable. This shows that increasing the learning rate can speed up optimization only up to a point; when the learning rate is too large, SGD overshoots the minimum and becomes unstable.
